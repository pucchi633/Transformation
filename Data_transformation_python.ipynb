#Q7. Build Preprocessing Pipeline

Create a pipeline that:

Applies encoding to categorical columns

Applies scaling to numerical columns

Use:

ColumnTransformer

Pipeline


```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Categorical and Numerical columns
cat_cols = ['Work_Mode_Hybrid']
num_cols = ['Hours_Worked_Per_Week', 'Salary_Standard']

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), cat_cols),
        ('num', StandardScaler(), num_cols)
    ]
)

# Pipeline
pipeline = Pipeline([
    ('preprocessor', preprocessor)
])

# Apply pipeline
result = pipeline.fit_transform(df)

print(result)
```

#Q8. Apply Pipeline

Apply the pipeline on the dataset.

Display:

Transformed dataset

Shape of final dataset


```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
# Categorical and Numerical columns
cat_cols = ['Work_Mode_Hybrid']
num_cols = [ 'Hours_Worked_Per_Week', 'Salary_Standard']

# Column Transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), cat_cols),
        ('num', StandardScaler(), num_cols)
    ]
)

# Pipeline
pipeline = Pipeline([
    ('preprocessor', preprocessor)
])

# Apply Pipeline
transformed_data = pipeline.fit_transform(df)

# Display transformed dataset
print(transformed_data)

# Shape of final dataset
print("Shape:", transformed_data.shape)



transformed_df = pd.DataFrame(transformed_data)

transformed_df.head()

print("Shape:", transformed_df.shape)
```

# Q9. Why is scaling important in Python?


Scaling is an important data preprocessing step in Python, especially when working with Machine Learning.

Scaling means converting numerical features to a similar range or scale, so that one feature does not unfairly dominate another.


Why is scaling important?
#Prevents large-valued features from dominating
A feature with values in thousands shouldn't automatically have more influence than one with values between 0 and 100.

#Improves distance-based algorithms
Scaling is particularly important for:
K-Nearest Neighbors (KNN)
K-Means Clustering
Support Vector Machines (SVM)

#Helps optimization converge faster
Algorithms such as Linear Regression, Logistic Regression, and Neural Networks can benefit from appropriately scaled features.

#Makes features comparable
Different units such as age, income, weight, and treatment cost can be brought to a common scale.
